```python
# instalar librerias

import pandas as pd
import openpyxl
from openpyxl.styles import Font, PatternFill, Alignment, Border, Side
from openpyxl.utils import get_column_letter
from google.colab import files
import io, os

 # configuración de las 4 hojas que integran el archivo

HOJAS = [
    {
        "nombre": "1. Información apoyos administr",
        "fila_inicio": 3,
        "columnas": list(range(0, 8)),    # A:H
        "n_cols": 8,
    },
    {
        "nombre": "2. Viajeros Frecuentes",
        "fila_inicio": 3,
        "columnas": list(range(0, 10)),   # A:J
        "n_cols": 10,
    },
    {
        "nombre": "3. Herramienta de Autogestión ",
        "fila_inicio": 3,
        "columnas": list(range(0, 13)),   # A:M
        "n_cols": 13,
    },
    {
        "nombre": "4. Recomendaciones ",
        "fila_inicio": 3,
        "columnas": list(range(0, 3)),    # A:C
        "n_cols": 3,
    },
]

ARCHIVO_SALIDA = "MAESTRO_CONSOLIDADO.xlsx"

print("✅ Configuración cargada correctamente.")

```

```python
# Cargue del archivo y dependecias desde google drive

print("📂 Selecciona TODOS los archivos .xlsx de las dependencias:")
archivos_subidos = files.upload()   # Abre el selector de archivos

nombres = list(archivos_subidos.keys())
print(f"\n✅ {len(nombres)} archivo(s) cargado(s):")
for n in nombres:
    print(f"   • {n}")


```

```python
# formato de las hojas

def borde_fino():
    lado = Side(style="thin", color="BFBFBF")
    return Border(left=lado, right=lado, top=lado, bottom=lado)

def estilo_encabezado(cell):
    cell.font    = Font(bold=True, color="FFFFFF", name="Arial", size=10)
    cell.fill    = PatternFill("solid", start_color="1F4E79")
    cell.alignment = Alignment(horizontal="center", vertical="center", wrap_text=True)
    cell.border  = borde_fino()

def estilo_datos(cell):
    cell.font      = Font(name="Arial", size=9)
    cell.alignment = Alignment(vertical="center", wrap_text=True)
    cell.border    = borde_fino()

def estilo_origen(cell):
    cell.font      = Font(bold=True, name="Arial", size=9, italic=True)
    cell.fill      = PatternFill("solid", start_color="D6E4F0")
    cell.alignment = Alignment(vertical="center")
    cell.border    = borde_fino()

def leer_hoja(contenido_bytes, nombre_archivo, cfg):
    """Lee encabezados y datos de una hoja desde bytes en memoria."""
    try:
        buffer = io.BytesIO(contenido_bytes)
        df_header = pd.read_excel(
            buffer, sheet_name=cfg["nombre"],
            header=None, skiprows=1, nrows=1,
            usecols=cfg["columnas"], dtype=str,
        )
        buffer.seek(0)
        df_data = pd.read_excel(
            buffer, sheet_name=cfg["nombre"],
            header=None, skiprows=cfg["fila_inicio"] - 1,
            usecols=cfg["columnas"], dtype=str,
        )
        df_data.dropna(how="all", inplace=True)
        df_data.reset_index(drop=True, inplace=True)
        return df_header, df_data
    except Exception as e:
        print(f"  ⚠  '{cfg['nombre']}' en {nombre_archivo}: {e}")
        return None, None

print("✅ Funciones auxiliares definidas.")
```

```python
# ── CELDA DE DIAGNÓSTICO: Ver nombres reales de hojas ──────────────
import io

print("📋 Nombres EXACTOS de hojas encontradas en cada archivo:\n")
for nombre_archivo, contenido in archivos_subidos.items():
    buffer = io.BytesIO(contenido)
    xl = pd.ExcelFile(buffer)
    print(f"📄 {nombre_archivo}")
    for i, hoja in enumerate(xl.sheet_names):
        print(f"   [{i}] '{hoja}'")
    print()
```

```python
# Consolidación  y generación de la información en un solo archivo maestro
# Acumulador por hoja

datos_por_hoja = {
    cfg["nombre"]: {"encabezados": None, "bloques": []}
    for cfg in HOJAS
}

# ── Leer todos los archivos subidos ─────────────────────────────────
for nombre_archivo, contenido in archivos_subidos.items():
    nombre_dep = os.path.splitext(nombre_archivo)[0]
    print(f"\n🔄 Procesando: {nombre_dep}")

    for cfg in HOJAS:
        df_header, df_data = leer_hoja(contenido, nombre_archivo, cfg)
        if df_data is None or df_data.empty:
            print(f"     ↳ Sin datos en '{cfg['nombre']}'")
            continue

        hoja_info = datos_por_hoja[cfg["nombre"]]
        if hoja_info["encabezados"] is None and df_header is not None:
            hoja_info["encabezados"] = df_header

        hoja_info["bloques"].append({"dependencia": nombre_dep, "datos": df_data})
        print(f"     ↳ '{cfg['nombre']}': {len(df_data)} filas leídas")

# ── Construir el workbook ────────────────────────────────────────────
wb = openpyxl.Workbook()
wb.remove(wb.active)

resumen_filas = []

for cfg in HOJAS:
    hoja_info = datos_por_hoja[cfg["nombre"]]
    ws = wb.create_sheet(title=cfg["nombre"])
    n_cols = cfg["n_cols"]
    fila_actual = 1

    # Título (fila 1)
    ws.merge_cells(start_row=1, start_column=1, end_row=1, end_column=n_cols)
    t = ws.cell(row=1, column=1, value=cfg["nombre"])
    t.font      = Font(bold=True, color="FFFFFF", name="Arial", size=11)
    t.fill      = PatternFill("solid", start_color="0D2D4E")
    t.alignment = Alignment(horizontal="center", vertical="center")
    ws.row_dimensions[1].height = 22
    fila_actual = 2

    # Encabezados (fila 2)
    if hoja_info["encabezados"] is not None:
        for ci, val in enumerate(hoja_info["encabezados"].iloc[0], start=1):
            c = ws.cell(row=fila_actual, column=ci,
                        value=str(val) if pd.notna(val) else "")
            estilo_encabezado(c)
        ws.row_dimensions[fila_actual].height = 30
        fila_actual += 1

    # Datos por dependencia
    total_registros = 0
    for bloque in hoja_info["bloques"]:
        dep = bloque["dependencia"]
        df  = bloque["datos"]

        # Fila separadora con nombre de dependencia
        ws.merge_cells(start_row=fila_actual, start_column=1,
                       end_row=fila_actual, end_column=n_cols)
        c = ws.cell(row=fila_actual, column=1, value=f"▶  {dep}")
        estilo_origen(c)
        ws.row_dimensions[fila_actual].height = 16
        fila_actual += 1

        # Registros
        for _, row in df.iterrows():
            for ci, val in enumerate(row, start=1):
                c = ws.cell(row=fila_actual, column=ci,
                            value="" if pd.isna(val) else val)
                estilo_datos(c)
            fila_actual += 1
            total_registros += 1

    resumen_filas.append({
        "Hoja": cfg["nombre"],
        "Registros": total_registros,
    })

    # Ajustar ancho de columnas
    for col_idx in range(1, n_cols + 1):
        max_len = 0
        for row in ws.iter_rows(min_col=col_idx, max_col=col_idx):
            for cell in row:
                try:
                    max_len = max(max_len, len(str(cell.value or "")))
                except:
                    pass
        ws.column_dimensions[get_column_letter(col_idx)].width = min(max(max_len + 2, 12), 45)

    ws.freeze_panes = "A3"
    print(f"\n✅ Hoja '{cfg['nombre']}': {total_registros} registros consolidados")

# ── Hoja RESUMEN ─────────────────────────────────────────────────────
ws_r = wb.create_sheet(title="RESUMEN", index=0)
ws_r.sheet_properties.tabColor = "FF6600"

ws_r.merge_cells("A1:C1")
t = ws_r.cell(row=1, column=1, value="RESUMEN DE CONSOLIDACIÓN")
t.font      = Font(bold=True, color="FFFFFF", name="Arial", size=12)
t.fill      = PatternFill("solid", start_color="0D2D4E")
t.alignment = Alignment(horizontal="center", vertical="center")
ws_r.row_dimensions[1].height = 25

for ci, h in enumerate(["Hoja", "Archivos fuente", "Registros consolidados"], 1):
    c = ws_r.cell(row=2, column=ci, value=h)
    estilo_encabezado(c)

for ri, fila in enumerate(resumen_filas, start=3):
    ws_r.cell(row=ri, column=1, value=fila["Hoja"]).border   = borde_fino()
    ws_r.cell(row=ri, column=2, value=len(nombres)).border   = borde_fino()
    ws_r.cell(row=ri, column=3, value=fila["Registros"]).border = borde_fino()
    for ci in range(1, 4):
        ws_r.cell(row=ri, column=ci).font = Font(name="Arial", size=9)

ws_r.column_dimensions["A"].width = 38
ws_r.column_dimensions["B"].width = 18
ws_r.column_dimensions["C"].width = 24

# ── Guardar en memoria y descargar ───────────────────────────────────
buffer_salida = io.BytesIO()
wb.save(buffer_salida)
buffer_salida.seek(0)

with open(ARCHIVO_SALIDA, "wb") as f:
    f.write(buffer_salida.read())

print(f"\n🎉 Consolidación completa. Iniciando descarga de '{ARCHIVO_SALIDA}'...")





```

```python
# Descargar el archivo maestro

files.download(ARCHIVO_SALIDA)
print("✅ Descarga iniciada.")
```
